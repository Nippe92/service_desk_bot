```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
import joblib 

df = pd.read_csv('../data/all_tickets_processed_cleaned.csv')

vectorizer = TfidfVectorizer(max_features=5000, stop_words='english')


X = vectorizer.fit_transform(df['cleaned_text'])

print(f"TF-IDF-matrisens form: {X.shape}")

joblib.dump(vectorizer, '../model/tfidf_vectorizer.joblib')

print("Vectorizer sparad till ../model/tfidf_vectorizer.joblib")
```

train och test split

```python
from sklearn.model_selection import train_test_split

y = df['Topic_group']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,      
    random_state=42
)

print(f"Träningsdata: {X_train.shape}, Testdata: {X_test.shape}")
```

```python
joblib.dump((X_train, X_test, y_train, y_test), "../data/train_test_split.joblib")
```
